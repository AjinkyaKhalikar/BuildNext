```python
import os
import re
import json
from dotenv import load_dotenv
from google import genai

load_dotenv()
API_KEY = os.getenv("GEMINI_API_KEY")
```

```python
with open("python_projects.json") as f:
    data = json.load(f)

projects = data['projects']

def get_features():
    difficulty_level = input("Input difficulty level for Easy\nMedium\nHard\nExpert")

    suggestion_type = input("Do you want weekend project or learning based project. Enter WEP for weekend and LBP for learning based project")

    return difficulty_level.capitalize(), suggestion_type.lower()
```

```python
get_features()
```

```python
def get_suggestion(difficulty):
    suggestions = []

    for project in projects:
        if project["difficulty_label"] != difficulty:
            continue
    
    for suggest in suggestions:
        print(f"\nProject id: {suggest['id']}\nTitle: {suggest['title']}\nDescription: {suggest['description']}\nPre-requists:{suggest['prerequisite_skills']}\nEstimated hours:{suggest['estimated_hours']}\n")
    proj_id = input("Which project you would like to choose please enter its id")

    return proj_id
```

```python
client = genai.Client(api_key=API_KEY)

for model in client.models.list():
    print(model.name)
```

```python
def ai_tutor(proj_id):
    SYSTEM_PROMPT = """
        You are an expert programming mentor.

        You help developers complete projects.

        Rules:
        - Never give complete code.
        - Give hints.
        - Explain concepts.
        - Help build logic.
        - Use pseudocode if necessary.
        - Ask questions that help the learner think.
    """
    proj_details = next(p
                         for p in projects 
                         if p['id'] == proj_id)

    proj_details = json.dumps(proj_details,indent=2)

    prompt = f"""{SYSTEM_PROMPT}

                Project Information:
                {proj_details}"""
    
    client = genai.Client(api_key= API_KEY)
    chat = client.chats.create(model= 'gemini-3.5-flash')

    chat.send_message(prompt)

    while True:
        question = input("You: ")

        if question == "exit":
            break

        response = chat.send_message(question)

        print(response.text)
```
